```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
data = {
    "tweet": [
        "I just got a promotion! 🎉",
        "Feeling down today... 😢",
        "That was so rude! 😠",
        "What a surprise party!! 😲",
        "Enjoying the sunshine ☀️",
        "Lost my keys again... 😞",
        "Best day of my life! 🥳",
        "I can't believe this happened 😤",
        "This food is amazing! 😊",
        "I am so tired and sad today 😔",
        "Just won a game! So happy! 🏆",
        "This is absolutely disgusting! 😡",
        "Did not expect this at all! 😮",
        "Feeling grateful and joyful 🌟",
        "Everything is going wrong today 😟"
    ],
    "label": [
        "happy",
        "sad",
        "angry",
        "surprise",
        "happy",
        "sad",
        "happy",
        "angry",
        "happy",
        "sad",
        "happy",
        "angry",
        "surprise",
        "happy",
        "sad"
    ]
}

df = pd.DataFrame(data)
```

```python
df.to_csv("tweet_emotion_dataset.csv", index=False)
print("Dataset saved as 'tweet_emotion_dataset.csv'")
```

```python
df = pd.read_csv("tweet_emotion_dataset.csv")

print("\n" + "="*55)
print("   CUSTOM TWEET EMOTION BENCHMARK DATASET")
print("="*55)
print(df.to_string(index=True))
```

```python
train, test = train_test_split(df, test_size=0.2, random_state=42)

print("\n" + "="*55)
print("   TRAINING SET")
print("="*55)
print(train.to_string(index=True))

print("\n" + "="*55)
print("   TESTING SET")
print("="*55)
print(test.to_string(index=True))

print(f"\n  Total Samples  : {len(df)}")
print(f"  Training Size  : {len(train)}")
print(f"  Testing Size   : {len(test)}")
```

```python
label_counts = df["label"].value_counts()

plt.figure(figsize=(8, 5))
bars = plt.bar(
    label_counts.index,
    label_counts.values,
    color=["skyblue", "salmon", "lightgreen", "orange"]
)

# Add count labels on top of each bar
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.1,
        str(int(bar.get_height())),
        ha="center",
        va="bottom",
        fontsize=12,
        fontweight="bold"
    )

plt.title("Emotion Label Distribution in Tweet Dataset", fontsize=14)
plt.xlabel("Emotion",  fontsize=12)
plt.ylabel("Count",    fontsize=12)
plt.tight_layout()
plt.savefig("emotion_distribution.png")
plt.show()

print("\nChart saved as 'emotion_distribution.png'")
print("\n" + "="*55)
print("   Custom Benchmark Dataset Created Successfully!")
print("="*55)
```
