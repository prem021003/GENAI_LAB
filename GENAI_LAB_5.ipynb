```python
def read_medical_reports(directory):
    reports = []

    for filename in os.listdir(directory):
        if filename.endswith(".csv"):
            filepath = os.path.join(directory, filename)

            with open(filepath, "r", encoding="utf-8") as file:
                reader = csv.reader(file)
                lines = list(reader)

                patient_name = lines[0][1].strip()
                report_date  = lines[1][1].strip()
                test_data    = lines[3:]

                test_results = []
                for row in test_data:
                    if len(row) >= 4:
                        test_results.append({
                            "Test"            : row[0].strip(),
                            "Result"          : row[1].strip(),
                            "Unit"            : row[2].strip(),
                            "Reference Range" : row[3].strip()
                        })

                reports.append({
                    "File"         : filename,
                    "Patient Name" : patient_name,
                    "Report Date"  : report_date,
                    "Tests"        : test_results
                })

    return reports

directory = "medical_reports"
reports   = read_medical_reports(directory)

for report in reports:
    print(f"\n{'='*45}")
    print(f"  Report File : {report['File']}")
    print(f"  Patient     : {report['Patient Name']}")
    print(f"  Date        : {report['Report Date']}")
    print(f"{'='*45}")
    print(f"  {'Test':<20} {'Result':<10} {'Unit':<12} {'Ref Range'}")
    print(f"  {'-'*55}")
    for test in report["Tests"]:
        print(f"  {test['Test']:<20} {test['Result']:<10} {test['Unit']:<12} {test['Reference Range']}")
```
