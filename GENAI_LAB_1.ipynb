```python
!pip install -q transformers ipywidgets

from transformers import pipeline
from ipywidgets import interact, FloatSlider, IntSlider, Checkbox
import torch

generator = pipeline("text-generation", model="gpt2")

def generate_text(prompt, temperature=0.7, top_p=0.9, max_tokens=50, deterministic=False):
    if deterministic:
        output = generator(
            prompt,
            max_new_tokens=max_tokens,
            do_sample=False,
            pad_token_id=generator.tokenizer.eos_token_id
        )
    else:
        output = generator(
            prompt,
            max_new_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            do_sample=True,
            pad_token_id=generator.tokenizer.eos_token_id
        )
    return output[0]["generated_text"]

@interact(
    prompt="The future of AI is",
    temperature=FloatSlider(min=0.1, max=2.0, step=0.1, value=0.7, description="Temperature"),
    top_p=FloatSlider(min=0.1, max=1.0, step=0.1, value=0.9, description="Top P"),
    max_tokens=IntSlider(min=20, max=200, step=10, value=50, description="Max Tokens"),
    deterministic=Checkbox(value=False, description="Deterministic")
)
def interactive_text_generation(prompt, temperature, top_p, max_tokens, deterministic):
    result = generate_text(prompt, temperature, top_p, max_tokens, deterministic)
    print("\nGenerated Text:\n")
    print(result)
```
