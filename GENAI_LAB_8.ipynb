```python
from transformers import pipeline

# Load pre-trained text generation model
generator = pipeline(
    "text-generation",
    model="gpt2",
    temperature=0.7,
    top_p=0.9,
    max_new_tokens=250
)

def evaluate_rag(context, answer):
    prompt = f"""
Evaluate the following answer based on the RAG-triad framework:

1. Context Relevance - Is the answer relevant to the provided context?
2. Groundedness      - Is the answer based on factual information from the context?
3. Answer Relevance  - Does the answer directly address the question or request?

Context: {context}
Answer : {answer}

Provide a score out of 10 for each criterion with brief explanations.
"""
    output = generator(
        prompt,
        do_sample=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )
    return output[0]['generated_text']

# Test Case 1: Good Context + Relevant Answer
print("="*60)
print("        RAG-TRIAD EVALUATION TOOL")
print("="*60)

print("\n--- TEST CASE 1: GOOD CONTEXT + RELEVANT ANSWER ---\n")
context1 = "Patient A: Age 45, diagnosed with hypertension. Prescribed Lisinopril 10mg daily. BP: 140/90 mmHg."
answer1  = "Patient A is taking Lisinopril 10mg daily for hypertension."
print(f"Context: {context1}")
print(f"Answer : {answer1}")
print("\n--- Evaluation Result ---\n")
print(evaluate_rag(context1, answer1))

# Test Case 2: Poor Context + Irrelevant Answer
print("\n" + "="*60)
print("--- TEST CASE 2: POOR CONTEXT + IRRELEVANT ANSWER ---\n")
context2 = "Patient A: Age 45, diagnosed with hypertension."
answer2  = "The patient has diabetes and is taking insulin."
print(f"Context: {context2}")
print(f"Answer : {answer2}")
print("\n--- Evaluation Result ---\n")
print(evaluate_rag(context2, answer2))

print("\n" + "="*60)
print("  Note: Good context produces more accurate evaluations.")
print("="*60)
```
