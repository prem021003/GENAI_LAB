```python
file_path = "/content/medical_questions.xlsx"

def add_medical_data():
    try:
        df = pd.read_excel(file_path, engine="openpyxl")
    except FileNotFoundError:
        df = pd.DataFrame(columns=["id", "question", "symptoms", "diagnosis", "date_created"])

    if not df.empty:
        next_id = df["id"].max() + 1
    else:
        next_id = 1

    while True:
        question = input("Enter medical question (or type 'exit' to stop): ")
        if question.lower() == "exit":
            break

        symptoms  = input("Enter symptoms: ")
        diagnosis = input("Enter diagnosis: ")
        date_created = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        new_entry = pd.DataFrame({
            "id":           [next_id],
            "question":     [question],
            "symptoms":     [symptoms],
            "diagnosis":    [diagnosis],
            "date_created": [date_created]
        })

        df = pd.concat([df, new_entry], ignore_index=True)
        next_id += 1

    df.to_excel(file_path, engine="openpyxl", index=False)
    print(f"\nData saved to {file_path}\n")

def view_data():
    try:
        df = pd.read_excel(file_path, engine="openpyxl")
        print("\nStored Medical Questions:\n")
        print(df.to_string(index=False))
    except FileNotFoundError:
        print("\nNo data found. Please add some questions first.\n")
```

```python
# Main Execution
add_medical_data()
view_data()
```
